### From Text to Audio and Images

We've focused on **text-based LLMs** throughout this course. Why? Because we started with an AI assistant that understood our instructions (in text) and wrote Python code (in text).

And we continued to use text because it is at the core of many interesting problems.

But today's AI systems are **multimodal**. They can handle (i.e., take in and produce):

-   text
-   audio
-   images
-   videos

QUESTION: How do multimodal AI systems work?

----

### Everything is a Number

Every computer file is a collection of numbers.

```python
hexdump -d YOUR_FILE | less
```

These numbers are stored as bits, a base-2 representation.

-   The base-10 number `5` in base-2 is `101` (i.e., 1 * 2^2 + 0 * 2^1 + 1 * 2^0).

-   It is easy to create a physical representation of a binary number, like `101`, using the presence and absence of something (e.g., charge/no charge on a line of capacitors or a hole/no hole in strip of paper). One state represents a `0` and the other a `1`.

-   The trick to making computation reliable on binary numbers is to make a computational unit output better 0s and 1s than it got as inputs.

Computers (and humans) don't think of a computer file as a collection of numbers, but as text, audio, images, or videos. Even text files are often interpreted not as simple text, but as a Python program, a Markdown file, or an interactive Python notebook.

> This is the computer (and us) **interpreting** a file of numbers as something specific.

----

### Every Computation is Math on Numbers

Now let us think of an AI system as a piece of computation.

-   It reads its inputs (files of numbers) under a particular interpretation.
-   It produces its outputs (also files of numbers) and expects you to interpret them in a particular manner.

But during the computation, the AI system transforms the input into another form for its own purposes.

### Inside an AI

Inside an AI system, you'll find a **neural network**, which is just a computation on the binary numbers in the inputs.

No matter what the actual input, these systems convert the input into **tokens** or **vectors in a shared vector space**, which are just numbers or lists of numbers. Processing the different modalities typically starts by:

-   **Text:** Breaking the text into subword tokens.
-   **Images:** Splitting an image (just numbers) into a grid of small pixel squares.
-   **Audio:** Slicing the time-based waveform (just numbers) into short windows.
-   **Video:** Working with the audio and image streams underlying the video.

Next, the systems align the encoded multimodal data so similar concepts appear close to each other in the vector space. For example, the following would all appear near each other:

-   The word "dog".
-   The sound of a dog barking.
-   An image of a dog.

> To go deeper, take a class or read [Maarten Grootendorst and Jay Alammar's _A Visual Guide to Multimodal Generation_](https://newsletter.maartengrootendorst.com/p/a-visual-guide-to-multimodal-generation).

### Voice Memos

**Problem statement:** I am constantly on the go and often think of something that I need to remember to do later. It would be convenient to record a voice memo, but not to leave it in that form.

I'd like a tool that would:

-   Read the voice memo (maybe even capture it so I could use just one tool).
-   Transcribe it, i.e., create a text note.
-   Take the text note and:
    - add the items into my to-do list, or
    - send them to me in an email.

**Let's decompose the task.** What's a first good step?

- Let's take an existing voice recording, send it to an AI model capable of analyzing audio, and ask it to respond with structured JSON.
    
- See `vmemo1.py`, which illustrates how you can put the audio file directly into the prompt input. It uses base64 encoding and [Pydantic validation](https://pydantic.dev/docs/validation/latest/get-started/) to create a schema for the AI model's response.

- You've learned how to work with a new datatype, i.e., something that isn't text.

**Improve the prototype.** Eliminate the separate app for taking a voice memo and the hassle of pointing the program at an audio file.

-   See `vmemo2.py`, which adds the function `record_memo`, saves the recorded memo as a WAV file, and strips out the code that handled the many different audio formats.

**Complete the prototype.** Take the text extracted from the audio and send an email.

-   I prefer email, but you could use anything (a cloud-based to-do list, a text message, etc.)

-   See `vmemo3.py`, which hardcodes the email recipient and adds a new function `email_notes`.

-   The program `vmemo3_test.py` allows you to test this new function without having to record a new memo or invoke an AI to transcribe it.

### Reusable Lessons

1.  Start your project with a personal pain point.
    - Feel free to combine your desire to learn something new with AI suggestions on what projects you might build while learning.

2.  Decompose your first idea and focus on one narrow job.

3.  Build the smallest Python prototype, which probably isn't a web app.

4.  Add a human review step.

5.  Test failures as well as successes.

6.  Discuss practical boundaries. What should an app do when it's uncertain?

7.  Add features and complexity a step at a time.
